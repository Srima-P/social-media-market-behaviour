```python
from transformers import pipeline

finbert = pipeline(
    "sentiment-analysis",
    model="ProsusAI/finbert",
    truncation=True,
    max_length=512
)
```

```python
finbert("GME earnings exceeded expectations")
```

```python
finbert("AMC reported disappointing results")
```

```python
import pandas as pd
gme_df = pd.read_csv("reddit_gme.csv")
amc_df = pd.read_csv("reddit_amc.csv")
tsla_df = pd.read_csv("reddit_tsla.csv")
```

```python
gme_df["text"].str.len().describe()
```

```python
long_posts = (gme_df["text"].str.len() > 2000).sum()
print(long_posts)
```

```python
print(
    round(
        long_posts / len(gme_df) * 100,
        2
    ),
    "%"
)
```

```python
gme_df["text"].str.len().quantile([0.90, 0.95, 0.99])
```

```python
for post in gme_df["text"].sample(20):
    result = finbert(post)
    print(result)
```

```python
sample_results = []

for post in gme_df["text"].head(100):
    sample_results.append(finbert(post))
```

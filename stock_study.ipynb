```python
import yfinance as yf
import pandas as pd

stocks = ["GME", "AMC", "TSLA"]

all_data = []

for stock in stocks:

    data = yf.download(
        stock,
        start="2021-01-01",
        end="2021-03-01",
        auto_adjust=False
    )

    # Remove MultiIndex columns
    if isinstance(data.columns, pd.MultiIndex):
        data.columns = data.columns.get_level_values(0)

    data = data.reset_index()

    data["Ticker"] = stock

    all_data.append(data)

market_data = pd.concat(all_data, ignore_index=True)

market_data.head()
```
